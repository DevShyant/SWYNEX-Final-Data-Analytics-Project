```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20, kde=True, color='skyblue')
plt.title('Age Distribution')
plt.show()
```

```python
import seaborn as sns
sns.countplot(x='Gender', data=df,hue='Gender',palette='Set2')
plt.title('Gender Distribution')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='Location', data=df, order=df['Location'].value_counts().index)
plt.xticks(rotation=45)
plt.title('Number of Purchases by Location')
plt.show()
```

```python
df['Online/Offline'].value_counts().plot.pie(autopct='%1.1f%%', colors=['lightgreen','lightcoral'])
plt.title('Online vs Offline Purchases')
plt.ylabel('')
plt.show()
```
